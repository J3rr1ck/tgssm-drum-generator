## 🎧 Interactive Multi-Genre Drum Synthesis Studio
Type any descriptive text prompt and click **Generate Drum Sample** to synthesize and audition 44.1kHz studio-grade drum waveforms in real time!

```python
# 4. Interactive Synthesis Studio Widget
import ipywidgets as widgets
from IPython.display import display, Audio
import soundfile as sf

prompt_box = widgets.Text(
    value="808, sub kick, sub bass, deep sub, deep 50Hz sub, heavy low end, hard transient click",
    description="Drum Prompt:",
    layout=widgets.Layout(width="80%")
)
steps_slider = widgets.IntSlider(value=30, min=15, max=50, step=5, description="ODE Steps:")
cfg_slider = widgets.FloatSlider(value=3.0, min=1.0, max=6.0, step=0.5, description="CFG Scale:")
gen_button = widgets.Button(description="🥁 Generate Drum Sample", button_style="success")
out_widget = widgets.Output()

def on_generate_clicked(b):
    with out_widget:
        out_widget.clear_output()
        prompt = prompt_box.value
        print(f"⚡ Synthesizing 44.1kHz waveform for: "{prompt}"")
        out_wav = "studio_sample.wav"
        !PYTHONPATH=python python3 python/generate_dac_drum.py --prompt "{prompt}" --output {out_wav} --steps {steps_slider.value} --cfg {cfg_slider.value}
        display(Audio(out_wav, rate=44100, autoplay=True))

gen_button.on_click(on_generate_clicked)
display(widgets.VBox([prompt_box, steps_slider, cfg_slider, gen_button, out_widget]))
```
